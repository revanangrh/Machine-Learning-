```python
# ============================================
# EVALUASI CLUSTERING K-MEANS DATASET IRIS
# GOOGLE COLAB
# ============================================

# Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.metrics import davies_bouldin_score

# ============================================
# LOAD DATASET IRIS
# ============================================

iris = load_iris()

# Data fitur
X = iris.data

# Menampilkan nama fitur
print("Nama Fitur:")
print(iris.feature_names)

# ============================================
# PROSES CLUSTERING K-MEANS
# ============================================

# Membuat model K-Means
kmeans = KMeans(n_clusters=3, random_state=42)

# Melatih model
kmeans.fit(X)

# Hasil cluster
y_kmeans = kmeans.predict(X)

# ============================================
# MENAMPILKAN HASIL CLUSTER
# ============================================

print("\nHasil Cluster:")
print(y_kmeans)

# ============================================
# EVALUASI CLUSTERING
# ============================================

# Silhouette Score
silhouette = silhouette_score(X, y_kmeans)

# Davies Bouldin Score
dbi = davies_bouldin_score(X, y_kmeans)

# Inertia
inertia = kmeans.inertia_

print("\n===== HASIL EVALUASI =====")
print("Silhouette Score :", silhouette)
print("Davies Bouldin Index :", dbi)
print("Inertia :", inertia)

# ============================================
# VISUALISASI CLUSTER
# ============================================

plt.figure(figsize=(8,6))

# Scatter plot cluster
plt.scatter(X[:,0], X[:,1],
            c=y_kmeans,
            cmap='viridis',
            s=50)

# Centroid cluster
centers = kmeans.cluster_centers_

plt.scatter(centers[:,0],
            centers[:,1],
            c='red',
            s=200,
            alpha=0.8,
            marker='X')

# Judul dan label
plt.title("Clustering K-Means Dataset Iris")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.show()

# ============================================
# KESIMPULAN OTOMATIS
# ============================================

print("\n===== KESIMPULAN =====")

if silhouette > 0.5:
    print("Clustering cukup baik karena Silhouette Score mendekati 1")
else:
    print("Clustering kurang optimal")

if dbi < 1:
    print("Cluster memiliki pemisahan yang baik")
else:
    print("Cluster masih kurang baik")

print("Program clustering berhasil dijalankan")
```
